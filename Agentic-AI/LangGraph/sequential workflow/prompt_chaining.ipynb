### importing libraries

```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

### loadind env file and instantiating openai model

```python
load_dotenv()

model=ChatOpenAI()
```

### create state 

```python
class BlogState(TypedDict):
    title : str
    outline : str
    content : str
```

### create nodes, graph, add nodes, add edges

#### creating nodes

```python
def create_outline(state : BlogState) -> BlogState:
    
    title=state['title']

    prompt=f'create the outline for the given title {title}'

    outline=model.invoke(prompt).content    

    state['outline']=outline

    return state

def create_blog(state : BlogState) -> BlogState :
    outline=state['outline']

    prompt=f"create the blog for given title {state['title']} with specific outline {outline}"

    blog=model.invoke(prompt).content

    state['content']=blog 

    return state
```

```python
graph=StateGraph(BlogState)

graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)

graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog',END)
```

```python
print(graph.compile().get_graph().draw_mermaid())
```

```python
initial_state={'title': 'Rise of AI in India'}

workflow=graph.compile()

final_state=workflow.invoke(initial_state)


print(final_state)
```
