## Importing libraries

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

### Simple Chain

```python

load_dotenv()

prompt=PromptTemplate(
    template='Generate five points about the topic: {topic}',
    input_variables=['topic']
)

model=ChatOpenAI()


parser=StrOutputParser()

# Create a chain that combines the prompt, model, and parser
chain=prompt | model | parser

# Invoke the chain with a specific topic
result=chain.invoke({'topic':'Time travel'})

#output
print('The result is: ',result)
```

#### visualize the chain

```python
chain.get_graph().print_ascii()
```

### Sequential chain 

```python
prompt1=PromptTemplate(
    template='Generate a report detailed report on the topic: \n {topic}',
    input_variables=['topic']
)

prompt2=PromptTemplate(
    template='Generate five line pointer summary for the given text: \n {text}',
    input_varibles=['text']
)

model=ChatOpenAI()

parser=StrOutputParser()

chain=prompt1 | model | parser | prompt2 | model | parser

result=chain.invoke({'topic':'motherhood'})

print('The result is: ',result)
```

```python
chain.get_graph().print_ascii()
```
