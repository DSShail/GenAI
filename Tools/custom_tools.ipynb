### Method 2: Using structured tool and pydantic

```python
from langchain.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a:int =Field(required=True,description="First number to add")
    b:int =Field(required=True,description="second number to add")
```

```python
def multiply_func(a:int,b:int)-> int:
    return a*b
```

```python
multiply_tool=StructuredTool.from_function(
    func=multiply_func,
    name='multiply',
    description="Multiply two numbers",
    args_schema=MultiplyInput
)
```

```python
result=multiply_tool.invoke({'a':3,'b':4})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
```

- Base tool : it is the abstract base class for all tools in Langchain. It defines the core structure and interface that any tool must follow, whether it's simple one-liner or a fully customized function
- other tools like @tool, structured tool are build on top of Basetool
